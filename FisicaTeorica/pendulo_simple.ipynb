Consideremos un péndulo simple formado por una masa puntual suspendida de una varilla rígida sin masa , cuyo extremo superior permaneze fijo. Como se mostrará acontinuación, las herramientas que nos proporciona la mecánica analítica nos conducen a una ecuación de movimiento de tipo No Lineal, para la cual es habitual suponer pequeños desplazamientos en torno al punto de equilibrio estable, con el fin de estudiar las pequeñas oscilaciones del sistema. Si bien se busca mantener este tipo de comportamiento oscilatorio del sistema, se incluyen dos términos adicionales que son responsables en cierta medida de la no-linealidad de la solucion, los cuales prceden de la respectiva expansión en series de potencias de la función seno que aparece en la ecuación de movimiento.
Bajo la suposicion de que los coeficientes  son pequeños, que difieren entre si en un solo orden de magnitud en pequeñez, nos apoyamos en el método de las aproximaciones sucesivas modificado, el cual resulta ser una herramienta útil para afrontar ecuaciones diferenciales con nolinealidades pequeñas y de solución periodicas.

#### Se construye el lagrangiano del sistema:

![](pendulo1.gif)


La energía cinética del sistema está dada como
$$
   T = \frac{1}{2}ml^2\dot{\theta}^2 \tag{1}
$$

con $m$ la masa de la partícula, $l$ la longitud del péndulo y $\theta$ la variable dinámica del sistema.

La energía potencial gravitacional está dada por:

$$
   U = -mgl\cos{\theta} \tag{2}
$$

De lo anterior, se tiene para el lagrangiano

$$
L = T-U
$$

$$
L = \frac{1}{2}ml^2\dot{\theta}^2+mgl\cos{\theta} \tag{3}
$$

De la ecuación de Euler-Lagrange, obtenemos la ecuación de movimiento

$$
\ddot{\theta} +\frac{g}{l}\sin\theta= 0 \tag{4}
$$
con  $\omega_0^2=\frac{g}{l}$ el termino asociado a la frecuencia natural del sistema bajo pequeñas oscilaciones.

### Solución Analítica
Considerando la función $\sin(\theta)$ en las proximidades del punto equilibrio estable $\theta_0 = 0$, esta se puede expandir en su serie de Taylor:

$$
\sin(\theta) = \theta -\frac{{\theta}^3}{3!}+\frac{{\theta}^5}{5!}+... \tag{5}
$$

Reemplazando en la ecuación diferencial, hasta el segundo termino no lineal relevante, se tiene que:
$$
\ddot{\theta} +\omega _0 ^2 \theta = \frac {\omega_0 ^2}{ 6}\theta^3 - \frac {\omega_0 ^2}{120}\theta^5
$$

De la cual solo tomamos los dos primeros términos no lineales,de modo que:

$$
\ddot{\theta} +\omega _0 ^2 \theta = \alpha \theta^3 + \beta \theta^5 \tag{6}
$$

Busquemos la solución en forma de una serie de aproximaciones sucesivas:

$$
\theta = \theta^{(0)} + \lambda\theta^{(1)} + \lambda^2\theta^{(2)}
$$

$$
\omega = \omega_0 + \lambda\omega^{(1)} + \lambda^2\omega^{(2)}
$$
siendo
$$
\boxed{\theta^{(0)}=a\cos(\omega t)} \tag{7}
$$

Sumando y restando: $\frac {\omega_0 ^2}{\omega }$ escribimos la ecuación $(6)$ de la forma
$$
\frac{\omega_0^2}{\omega^2} \ddot{\theta} +  \omega_0^2 \theta = \lambda\alpha\theta^3 + \lambda^2\beta\theta^5  - \ddot{\theta}  + \frac{\omega_0^2}{\omega^2} \ddot{\theta}  \tag{8}
$$

Se puede escribir:
$$
\frac {1}{\omega^2}= \frac {1}{(\omega_0 + \lambda\omega^{(1)} + \lambda^2\omega^{(2)})^{2} }=\frac {1}{\omega_0^{2}+\lambda^{2}(\omega^{(1)})^{2}+\lambda^{4}(\omega^{(2)})^{2}+2\lambda \omega_0\omega^{(1)}+2\lambda^{2}\omega_0\omega^{(2)}+2\lambda^{3}\omega^{(1)}\omega^{(2)}}
$$

$$
\frac {1}{\omega^2}= \frac {1}{\omega_0^2}\frac {1}{1+\frac{\lambda^{2}{(\omega^{(1)})^{2}}}{\omega_0^2}+\frac{\lambda^{4}(\omega^{(2)})^{2}}{\omega_0^2}+\frac{2\lambda \omega^{(1)}}{\omega_0}+\frac{2\lambda^{2}\omega^{(2)}}{\omega_0}+\frac {2\lambda^{3}\omega^{(1)}\omega^{(2)}}{\omega_0^2}}
$$

Usando la aproximación: $ (1+\epsilon)^{-1} = 1 - \epsilon + \epsilon^{2}$ y salvando los términos hasta el orden $\lambda^{2}$,con lo cual:

$$
\frac {1}{\omega^2}\approx \frac {1}{\omega_0^2}({1-\frac{\lambda^{2}{(\omega^{(1)})^{2}}}{\omega_0^2}-\frac{2\lambda \omega^{(1)}}{\omega_0}-\frac{2\lambda^{2}\omega^{(2)}}{\omega_0}+\frac{4\lambda^{2}(\omega^{(1)})^{2}}{\omega_0^2}})  \tag{9}
$$


Reemplazando $(9)$ en $(8)$ y escrbiendo solo hasta el orden $\lambda^{2}$, se tiene:

$$
({1-\frac{2\lambda \omega^{(1)}}{\omega_0}-\frac{2\lambda^{2}\omega^{(2)}}{\omega_0}+\frac{3\lambda^{2}(\omega^{(1)})^{2}}{\omega_0^2}})(\ddot{\theta}^{(0)} + \lambda\ddot{\theta}^{(1)} + \lambda^2\ddot{\theta}^{(2)})+ {\omega_0^2}(\theta^{(0)} + \lambda\theta^{(1)} + \lambda^2\theta^{(2)})=\lambda\alpha({\theta^{(0)}}^3 + 3\lambda{\theta^{(0)}}^2\theta^{(1)})+\lambda^2\beta{\theta^{(0)}}^5
$$

$$
\ddot{\theta}^{(0)} + \lambda\ddot{\theta}^{(1)} + \lambda^2\ddot{\theta}^{(2)}-\frac{2\lambda \omega^{(1)}}{\omega_0}\ddot{\theta}^{(0)}-\frac{2\lambda^2 \omega^{(1)}}{\omega_0}\ddot{\theta}^{(1)}-\frac{2\lambda^{2}\omega^{(2)}}{\omega_0}\ddot{\theta}^{(0)}+\frac{3\lambda^{2}{\omega^{(1)}}^{2}}{\omega_0^2}\ddot{\theta}^{(0)}+{\omega_0}^{2}{\theta}^{(0)}+\lambda{\omega_0}^{2}{\theta^{(1)}}+\lambda^{2}{\omega_0}^{2}{\theta}^{2}=\lambda\alpha{\theta^{(0)}}^3+3\lambda^2\alpha{\theta^{(0)}}^2\theta^{(1)}+\lambda^2\beta{\theta^{(0)}}^5
$$

Separando los coeficientes de las potencias iguales a $\lambda^k$

###### Orden $\lambda^0$

$$
\ddot{\theta}^{(0)}+{\omega_0}^{2}{\theta}^{(0)}=0
$$

$$
\boxed{{\theta}^{(0)}=a\cos({\omega_0}t)}
$$

###### Orden $\lambda^1$

$$
\ddot{\theta}^{(1)}-\frac{2\omega^{(1)}}{\omega_0}\ddot{\theta}^{(0)}+{\omega_0}^{2}{\theta}^{(1)}=\alpha{\theta^{(0)}}^3
$$

$$
\ddot{\theta}^{(1)}-\frac{2\omega^{(1)}}{\omega_0}[-a{\omega_0}^2\cos({\omega_0}t)]+{\omega_0}^{2}{\theta}^{(1)}=\alpha a^3\cos^3({\omega_0}t)
$$

$$
\ddot{\theta}^{(1)}+{\omega_0}^{2}{\theta}^{(1)}=\alpha a^3[\frac{3}{4}\cos({\omega_0}t)+\frac{1}{4}\cos({3\omega_0}t]-\frac{2\omega^{(1)}}{\omega_0}a{\omega_0}^2\cos({\omega_0}t)
$$

$$
\ddot{\theta}^{(1)}+{\omega_0}^{2}{\theta}^{(1)}=a(\frac{3}{4}\alpha a^2-2\omega_0\omega^{(1)})\cos({\omega_0}t)+\frac{1}{4}\alpha a^3\cos({3\omega_0}t)
$$

De modo que, se elimina el término resonante, escribiendo

$$
\frac{3}{4}\alpha a^2-2\omega_0\omega^{(1)}=0
$$
$$
\boxed{\omega^{(1)}=\frac{3}{8}\alpha\frac{a^2}{\omega_0}} \tag{10}
$$

Rescribiendo, se obtiene

$$
\ddot{\theta}^{(1)}+{\omega_0}^{2}{\theta}^{(1)}=\frac{1}{4}\alpha a^3\cos({3\omega_0}t)
$$

Cuya solución particular está dada como
$$
\boxed{{\theta}^{(1)}=-\frac{1}{32}\frac{\alpha a^3}{\omega_0^{2}}\cos({3\omega_0}t)}
$$

###### Orden $\lambda^2$

$$
\ddot{\theta}^{(2)}-\frac{2\omega^{(1)}}{\omega_0}\ddot{\theta}^{(1)}-\frac{2\omega^{(2)}}{\omega_0}\ddot{\theta}^{(0)}+\frac{3{\omega^{(1)}}^{2}}{\omega_0^2}\ddot{\theta}^{(0)}+{\omega_0}^{2}{\theta}^{(2)}=3\alpha{\theta^{(0)}}^2\theta^{(1)}+\beta{\theta^{(0)}}^5
$$

$$
\ddot{\theta}^{(2)}-\frac{2\omega^{(1)}}{\omega_0}\frac{9}{32}\alpha a^3\cos({3\omega_0}t)-\frac{2\omega^{(2)}}{\omega_0}(-{\omega_0}^{2}a)\cos({\omega_0}t)+\frac{3{\omega^{(1)}}^2}{{\omega_0}^2}({\omega_0}^{2}a)\cos({\omega_0}t)+{\omega_0}^{2}{\theta}^{(2)}=3\alpha a^{2}\cos^2({\omega_0}t)\frac{-1}{32}\frac{\alpha a^{3}}{{\omega_0}^{2}}\cos({3\omega_0}t)+\beta a^{3}\cos^3({\omega_0}t)
$$

$$
\ddot{\theta}^{(2)}+{\omega_0}^{2}{\theta}^{(2)}=\frac{9}{16}\frac{\omega^{(1)}}{\omega_0}\alpha a^3\cos({3\omega_0}t)-2\omega^{(2)}{\omega_0}^{2}a\cos({\omega_0}t)+3{\omega^{(1)}}^2a\cos({\omega_0}t)-\frac{3}{32}\frac{\alpha^2 a^{5}}{{\omega_0}^{2}}\cos^2({\omega_0}t)\cos({3\omega_0}t)+\beta a^{3}\cos^3({\omega_0}t)
$$

Lo cual se puede reescribir como

$$
\ddot{\theta}^{(2)}+{\omega_0}^{2}{\theta}^{(2)}=(-2\omega^{(2)}{\omega_0}^{2}a+\frac{27{\alpha^{2}}a^5}{64\omega_0^2}-\frac{3{\alpha^{2}}a^5}{128\omega_0^2}{{\beta}^{2}}+\frac{10}{16}\beta a^{5})\cos({\omega_0}t)+(\frac{27{\alpha^{2}}a^5}{128\omega_0^2}-\frac{3}{64}\frac{\alpha^2 a^{5}}{{\omega_0}^{2}}+\frac{5}{16}\beta a^{5})\cos({3\omega_0}t)+(-\frac{3}{64}\frac{\alpha^2 a^{5}}{{\omega_0}^{2}} + \frac{\beta a^5}{16})\cos({5\omega_0}t) \tag{11}
$$

Escogiendo ${\omega}^{(2)}$ para anular la resonancia, se tiene

$$
-2\omega^{(2)}{\omega_0} + \frac{51 \alpha^2 a^5}{128\omega_0^2} + \frac{10\beta a^4}{16}=0
$$

$$
\omega^{(2)}=\frac{1}{2\omega_0}(\frac{51 \alpha^2 a^5}{128\omega_0^2} + \frac{\beta a^4}{16})
$$


$$
\boxed{\omega^{(2)}=\frac{51 \alpha^2 a^5}{256\omega_0^3} + \frac{5\beta a^4}{16\omega_0}}
$$

Reemplazando $(10)$ en $(11)$ y considerando que ya no está presente el término de resonancia:

$$
\ddot{\theta}^{(2)}+{\omega_0}^{2}{\theta}^{(2)}=(\frac{21}{128}\frac{\alpha^2 a^{5}}{{\omega_0}^{2}}+\frac{5}{16}\beta a^{5})\cos({3\omega_0}t)+(\frac{-3\alpha^2 a^{5}}{64\omega_0^2}+\frac{\beta a^5}{16})\cos({5\omega_0}t)
$$

De modo que, finalmente se obtiene para ${\theta}^{(2)}$:

$$
\boxed{{\theta}^{(2)}=-\frac{1}{8{\omega_0}^{2}}(\frac{21}{128}\frac{\alpha^2 a^{5}}{{\omega_0}^{2}}+\frac{5}{16}\beta a^{5})\cos({3\omega_0}t)-\frac{1}{24{\omega_0}^{2}}(\frac{-3\alpha^2 a^{5}}{64\omega_0^2}+\frac{\beta a^5}{16})\cos({5\omega_0}t)}
$$

###### Comparación de la solución de las oscilaciones no lineales de un péndulo, obtenidas con diferentes órdenes de aproximación, aplicando una  solución numérica mediante las herrramientas de análisis numérico de  Python y  una aproximación analítica.

### Solución Analítica vs Solución Numérica
La siguiente gráfica muestra la solución del péndulo simple, con amplitud inicial pequeña y grande

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
# Odeint
from scipy.integrate import odeint

def solucion_analitica(amplitud,t,w0):
    a = amplitud*np.pi/180
    alpha = w0**2/6
    beta = -w0**2/120
    w1 = 3*alpha*a**2/8*w0
    w2 = (51*alpha**2*a**4/256*w0**3) + (5*beta*a**4/16*w0)
    w= w0+w1+w2
    theta0 = (a*(np.cos(w*t)))*180/np.pi
    theta1 = (-1*alpha*a**3*np.cos(3*w*t)/32*w0**2)*180/np.pi
    theta2 = (-1/(8*w0**2)*(21/128*a**5*alpha**2 + 5/16*a**5*beta)*np.cos(3*w*t) \
        -1/(24*w0**2)*((-3/(64*w0**2))*a**5*alpha**2 +1/16*a**5*beta)*np.cos(5*w*t))*180/np.pi
    theta_1 = theta0 + theta1
    theta_2 = theta0 + theta1 + theta2

    return theta0,theta_1,theta_2


# Intervalo de tiempo y otros parametros del sistema:
t = np.linspace(0,10,1000)
w0=1

#Definicion de la solcion numerica como una funcion que admite una ngulo inicial y un intervalo de tiempo

def dU_dt(U,t):
    return [U[1],-w0*np.sin(U[0])]

def solucion_numerica(amplitud,t):
    #condicion inical
    U0 = [amplitud*np.pi/180,0.]

    # Solución numérica
    US1 = odeint(dU_dt,U0,t)*180/np.pi

    US = []

    for i in range(len(t)):
        US.append([t[i],US1[i][0],US1[i][1]])

    US=np.array(US)
    U=np.transpose(US)
    return(U)

def grafica(amplitud):
    theta0,theta1,theta2 = solucion_analitica(amplitud,t,w0)
    U = solucion_numerica(amplitud,t)

    fig, axs = plt.subplots(2, 3, figsize=(12, 8))

    # Subplot 1
    axs[0, 0].plot(t, theta0,color='red', marker='o', label = 'Solución orden cero',markersize=3)
    axs[0, 0].plot(t, U[1],color='b', label = 'Solución numérica')
    axs[0, 0].set_title('Solución numérica $vs$ Solución orden cero')
    axs[0, 0].set_xlabel('t')
    axs[0, 0].set_ylabel(r'$\theta$ (grados)')
    axs[0, 0].legend()
    axs[0, 0].grid()

    # Subplot 2
    axs[0, 1].plot(t, theta1, color='green', marker='o', label = 'Solución primer orden',markersize=3)
    axs[0, 1].plot(t, U[1], color='m', label = 'Solución numérica')
    axs[0, 1].set_title('Solución numérica $vs$ Solución Primer Orden')
    axs[0, 1].set_xlabel('t')
    axs[0, 1].set_ylabel(r'$\theta$ (grados)')
    axs[0, 1].legend()
    axs[0, 1].grid()

    # Subplot 3
    axs[0, 2].plot(t, theta2, color='yellow', marker='o', label = 'Solución segundo orden',markersize=3)
    axs[0, 2].plot(t, U[1], color='c', label = 'Solución numérica')
    axs[0, 2].set_title('Solución numérica $vs$ Solución segundo orden')
    axs[0, 2].set_xlabel('t')
    axs[0, 2].set_ylabel(r'$\theta$ (grados)')
    axs[0, 2].legend()
    axs[0, 2].grid()

    # Subplot 4
    axs[1, 0].plot(t, (theta0-U[1]), color='r')
    axs[1, 0].set_title('Error')
    axs[1, 0].set_xlabel('t')
    axs[1, 0].set_ylabel(r'$\theta$ (grados)')
    axs[1, 0].grid()

    # Subplot 5
    axs[1, 1].plot(t, (theta1-U[1]), color='g')
    axs[1, 1].set_title('Error')
    axs[1, 1].set_xlabel('t')
    axs[1, 1].set_ylabel(r'$\theta$ (grados)')
    axs[1, 1].grid()

    # Subplot 6
    axs[1, 2].plot(t, (theta2-U[1]), color='c')
    axs[1, 2].set_title('Error')
    axs[1, 2].set_xlabel('t')
    axs[1, 2].set_ylabel(r'$\theta$ (grados)')
    axs[1, 2].grid()

    # Ajustar el espaciado entre subplots
    plt.tight_layout()

    # Mostrar el gráfico
    plt.show()
```

### Amplitud pequeña (10°)

```python
amplitud = 10
grafica(amplitud)

```

### Amplitud grande (110°)

```python
amplitud = 110
grafica(amplitud)
```

Como se puede observar de las gráficas anteriores, la aproximación analítica para la ecuación del sistema  es bastante cercana al comportamiento de la respectiva solución numérica cuando la amplitud inicial no se toma demasiado alejada del regimen lineal ( a lo  sumo de un ángulo de 10°). Sin embargo, para el ejemplo de un ángulo inicial muy grande, las diferencias ente las soluciones son más  notables, especialmente en el desfase temporal que presentan sus gráficos, apreciándose que el período de la ocilacion tiende a incrementarse con la amplitud inicial del péndulo de acuerdo al comportamiento de la solución numerica que podemos considerar como referencia, de modo que el método perturbativo falla estrepitosamente ante situaciones donde las oscilaciones dependen de la amplitud inicial.
